### Cell 5: Creating and inspecting lists
Creates a list, prints it, checks the type and length, and demonstrates slicing to obtain sublists.

```python
list = [7,8,9,4,5,6,1,2,3]
print(list)
print(type(list))
print(len(list))
print(list[:5])
print(list[5:])
```

Shows negative indexing to access elements from the end and `insert()` to add an element at a specific position.

```python
print(list[-4])
print(list)
list.insert(2, 7)
print(list)
```

### Cell 7: Appending items
Uses `append()` to add an item to the end of the list (mutates the list).

```python
list.append(100)
print(list)
```

### Cell 8: Item assignment by index
Demonstrates replacing an element by index to show lists are mutable.

```python
list[1] = 10
print(list)
```

### Cell 9: Slice assignment

### Cell 10: Removing by value
`remove(value)` deletes the first occurrence of the value from the list.

```python
list = [7,8,9,4,5,6,1,2,3]
print(list)
list.remove(3)
print(list)
```

### Cell 11: Removing when duplicates exist
`remove()` removes only the first matching element when duplicates are present.

```python
list = [7,8,9,3,4,5,6,1,2,3]
print(list)
list.remove(3)
print(list)
```

### Cell 12: Pop and returned value
`pop(index)` removes an item and returns it; useful when you need the removed value.

```python
list = [7,8,9,4,5,6,1,2,3]
print(list)
remove_item = list.pop(3)
print(list)
print("Removed item:", remove_item)
```

### Cell 13: Deleting a variable
`del var` removes a name from the namespace; accessing it afterwards raises a NameError.

```python
list2 = [7,8,9,3,4,5,6,1,2,3]
print(list2)
del list2
print(list2)
print(list2)
```

### Cell 14: Clearing a list
`clear()` empties the list but keeps the variable pointing to the same (now empty) object.

```python
my_list3 = [10, 20, 30]
print(my_list3)
my_list3.clear()
print(my_list3)
```

### Cell 15: Sort (ascending)
`list.sort()` sorts in-place in ascending order and mutates the original list.

```python
new_list = [10,1,2, 20, 30]
new_list.sort()
print(new_list)
```

### Cell 16: Sort (descending)
`list.sort(reverse=True)` sorts in descending order (in-place).

```python
new_list = [10,1,2, 20, 30]
new_list.sort(reverse=True)
print(new_list)
```

### Cell 17: Copying a list
`copy()` returns a shallow copy; subsequent mutations to the original do not affect the copy.

```python
lister = ["7", "12", "Apple", "Banana", "Cherry", "apple", "banana", "cherry"]
lis = lister.copy()
print(lister)
lister.append("20")
print(lister)
print(lis)
```

### Cell 18: Aliasing vs copying
Assigning a list to a new name (`b = a`) creates an alias; both names reference the same object.

```python
new_list = [10,1,2, 20, 30]
new_list_2 = new_list
print(new_list)
new_list.append(40)
print(new_list)
print(new_list_2)
```

### Cell 19: Concatenation (creates new list)
Using `+` combines lists into a new list without modifying the originals.

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
joined_list = list1 + list2
print(joined_list)
```

### Cell 20: Extend (in-place)
`list.extend(iterable)` appends elements of the iterable into the list, mutating it.

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list1.extend(list2)
print(list1)
```

### Cell 21: Reverse (in-place)
`list.reverse()` reverses the list order in-place.

```python
listings = [1,2,3,4,5,6]
listings.reverse()
print(listings)
```
