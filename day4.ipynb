Q. The factorial of a non negative integer (n) written as (n!) is the product of all positive integers less than or equal to n. By mathematical definition the factorial of 0 is 1, and factorials are not defined for negative numbers.

a. Write a program that ask the user to input a whole number and then print the factorial of that number. 


```python
num = int(input("Enter a whole number:  "))

f = 1
if num >= 0 :
    for x in range(1, num + 1):
        f*=x
else:
    print("Input must be positive")

print(f)
```

```python
inventory = {
    "Microchips": [50, 12.50],
    "Resistors": [500, 0.10],
    "Capacitors": [200, 0.45],
}

total_store_value = 0

for val in inventory.values():
    total_store_value += val[0]*val[1]
    
print(total_store_value)
```

Q. Take an integer input and reverse it

```python
num = int(input("Enter the num: "))
rev_num = 0

while num > 0:
    last_digit = num % 10
    rev_num = rev_num * 10 + last_digit
    num = num // 10


print(rev_num)
```

Q. In coordinate geometry analyzing a line segment between 2 points on a grid (x1, y1) and (x2,y2), often requires finding following things

    1. The distance: d = ((x2 - x1) ** 2 + (y2 - y1) ** 2)**0.5 
    2. The midpoint: m = ((x1+x2) / 2, (y1+y2) / 2)

Write the python function that accepts parameter x1, y1, x2, y2 and return 3 values,
 a. The distance
 b. The midpoint x
 c. The midpoint y

```python
def calc_distance_mid(x1, x2, y1, y2):    
    print("The distance:    ", ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5)
    print("The midpoint of x:    ", (x1 + x2) / 2)
    print("The midpoint of y:    ", (y2 - y1) / 2)
    
calc_distance_mid(78, 53, 25, 5)
```

Q. Write a function called power that accepts 2 params, base and exponent, and returns the base raise to the power of the exponent.

1. If the exponent is not provided it should default to 2.

```python
def power(base, exponent=2):
    return base ** exponent


print(power(2,2))
print(power(8,5))
print(power(10))
print(power(5))
```
