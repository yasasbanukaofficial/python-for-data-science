### Cell 1: Equality comparison
Demonstrates the equality operator `==` which checks whether two values are equal and returns a boolean.

```python
x = 5
y = 4
print(x==y)
```

### Cell 2: Logical AND
Shows how the `and` operator combines boolean expressions; both sides must be True for the result to be True.

```python
x = 10
y = 20

result = (x < 15) and (y < 15)
print(result)

result = (x < 15) and (y < 25)
print(result)
```

### Cell 3: Logical OR
Shows the `or` operator where the overall expression is True if at least one operand is True.

```python
x = 10
y = 20

result = (x < 15) or (y < 15)
print(result)

result = (x < 15) or (y < 25)
print(result)

result = (x < 5) or (y < 5)
print(result)
```

### Cell 4: Logical NOT
Demonstrates the `not` operator which inverts the truth value of an expression.

```python
x = 10

print(not x > 10)
```

```python
point = (0, 5)

match point:
	case (0, 0):
		print("Origin")
	case (0, y):
		print(f"On Y axis, y={y}")
	case (x, 0):
		print(f"On X axis, x={x}")
	case (x, y):
		print(f"Point: {x}, {y}")
```
