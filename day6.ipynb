Q. Write a python class name Circle. You're class should use a constant value for PI and require a radius to be provided when a new circle obj is instantiated
- You must include a constructor method to set up the necessary initial state
- Additionally, implement 2 instance method:
    1. calculate_circumference() -> To compute and print the circumference
    2. calculate_area() -> To compute and print the area

```python
import math
class Circle:   
    PI_VALUE = math.pi
    
    def __init__(self, radius):
        self.radius = radius
        
    def calculate_circumference(self):
        circumference = round(2 * pi * self.radius, 2)
        print(f"The circumference is {circumference}")
        
    
    def calculate_area(self):
        area = round(pi * self.radius ** 2, 2)
        print(f"The circumference is {area}")
        

circle1 = Circle(5)
circle1.calculate_circumference()
circle1.calculate_area()

circle2 = Circle(2.5)
circle2.calculate_circumference()
circle2.calculate_area()
```

Q. Write a python program that demonstrate inheritance by creating a base class named Employee and derived class Manager.
- The base class should require a name and base salary when a new object is instantiated and it must include a method to calculate and return the total wage.
- The derived class must inherit from the base class require an additional bonus amount upon instantiation, and properly adjust the pay calculation to include this bonus.

```python
class Employee:
    def __init__(self, name, base_salary):
        self.name = name
        self.base_salary = base_salary
        
    def total_wage(self):
        return self.base_salary
        
class Manager(Employee):
    def __init__(self, name, base_salary, bonus):
        super().__init__(name, base_salary)
        self.bonus = bonus

    def total_wage(self):
        return self.base_salary + self.bonus
    

emp = Employee("John Doe", 3000)
mgr = Manager("Jane Smith", 4000, 1500)

print(emp.name, "annual wage:", emp.total_wage())
print(mgr.name, "annual wage (with bonus):", mgr.total_wage())
print(isinstance(mgr, Employee))
```

```python
class Shapes:
    def area(self):
        raise NotImplementedError

class Square(Shapes):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side

class Triangle(Shapes):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

square = Square(4)
triangle = Triangle(6, 3)

shapes = [square, triangle]

for shape in shapes:
    print(type(shape).__name__, "area:", shape.area())
```

Q. Create a python program demonstrating inheritance and the use of arbitary keyword args (**kwargs).
- Define a base class, Shape whos constructor accepts (**kwargs) dynamically assign them as instant attributes.
- Create a sub class cylinder that inherits from shape. It's constructor should explicitly take radius and height as required parameters, while also accepting (**kwargs) for additional metadata
Example -> (unit, metadata) 
- Use super to pass the extra keyword argument to base class
- Implement a method calculate volume in cylinder that computes the volume of a cylinder

```python
import math

class Shape: 
    def __init__(self, **kwargs):
        for key, value in kwargs.items():
            setattr(self, key, value)
            

class Cylinder(Shape):
    def __init__(self, radius, height, **kwargs):
        self.radius = radius
        self.height = height
        super().__init__(**kwargs)
        
    def calculate_volume(self):
        return round(math.pi * self.radius ** 2, 3)
    
    
cylinder_1 = Cylinder(3, 5, unit="cm", color="blue")
cylinder_2 = Cylinder(7, 10, unit="m",material="steel")

print(f"Radius: {cylinder_1.radius}, height: {cylinder_1.height} with color of {cylinder_1.color}")
print(f"Radius: {cylinder_2.radius} and height {cylinder_2.height} with the material of {cylinder_2.material}")


print(cylinder_2.calculate_volume())
print(cylinder_1.calculate_volume())
```
