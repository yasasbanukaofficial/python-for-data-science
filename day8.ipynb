```python
arr_1 = np.array([20,24,16,8])
arr_2 = np.array([5,8,2,4])

list_1 = arr_1.tolist()
list_2 = arr_2.tolist()

list_sum = {
    "add": [a + b for a, b in zip(list_1, list_2)],
    "sub": [a - b for a, b in zip(list_1, list_2)],
    "mul": [a * b for a, b in zip(list_1, list_2)],
    "div": [a / b for a, b in zip(list_1, list_2)],
    "mod": [a % b for a, b in zip(list_1, list_2)],
}

print("List mathematical operations: ", list_sum)

print(arr_1 + arr_2)
print(arr_1 - arr_2)
print(arr_1 * arr_2)
print(arr_1 / arr_2)
print(arr_1 % arr_2)
```
