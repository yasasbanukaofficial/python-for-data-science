```python
url = "http://example.com"

res = requests.get(url)
match res.status_code:
    case 200:
        if "text/html" in res.headers['Content-Type']:
            soup = BeautifulSoup(res.text, 'html.parser')
            if not soup.h1.text:
                print('No valid H1 tag')
            else: print(soup.h1.text)
        else: 
            print("This file is not a html file")
    case 400:
        print("Not a valid url", res)
```

```python
params = {
	"latitude": 6.7132,
	"longitude": 79.9026,
	"current": ("temperature_2m", "wind_speed_10m"),
}
res = requests.get("https://api.open-meteo.com/v1/forecast", params=params)

match res.status_code:
    case 200:
        if "application/json" in res.headers['Content-Type']:
            resp = res.json()
            if resp is not None and 'current' in resp and 'current_units' in resp:
                    current = resp['current']
                    current_units = resp['current_units']
                    print("Temperature", current['temperature_2m'], current_units['temperature_2m'])
                    print("Wind Speed", current['wind_speed_10m'], current_units['wind_speed_10m'])
            else: print('There are no current keyword existing in this API check docs!')
        else:
            print("This doesn't give valid data")
    case 400:
        print("Error in fetching data")
```
