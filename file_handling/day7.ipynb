Q. A tution class maintains student attendance records using a text file name, attendance.txt.
 - Each line of the file contains a student's name and the number of attended sessions seperated by a comma (,)
 - Write a python program to:
    1. Add atleast 10 student records in to the file.
    2. Read all records from the file.
    3. Display the names of students who's attendance percentage is below 80%.
    4. Calculate and display the average attendance of the class.

```python

with open("attendance.txt", "w+") as file:
    TOTAL_SESSIONS = 30
    records = [
        ("Alice", 28),
        ("Bob", 24),
        ("Charlie", 18),
        ("Diana", 30),
        ("Eve", 27),
        ("Frank", 12),
        ("Grace", 20),
        ("Heidi", 29),
        ("Ivan", 15),
        ("Judy", 9),
    ]

    for name, attended in records:
        file.write(f"{name},{attended}\n")
        
    low_attendance = []
    total_attendance_percentage = total_student_count = 0
        
    file.seek(0)
    
    for l in file:
        line = file.readline()
        if not line:
            break
        
        name, count = line.strip().split(",")
        
        # Student ge percentage eka (individual)
        attendance_count = int(count)
        attendance_percentage = (attendance_count / TOTAL_SESSIONS) * 100
        
        # Total laminge attendance percentage eka
        total_attendance_percentage += attendance_percentage
        
        # Total laminge count eka
        total_student_count += 1
        
        if attendance_percentage < 80: 
            low_attendance.append((name, attendance_count))
            
    if total_student_count > 0:
        # Total lamainge avg eka
        class_avg = total_attendance_percentage / total_student_count
    else:
        class_avg = 0
        
    print(low_attendance)
    print(f"{class_avg:.2f}")
    
```

Q. A school store student examination records in a csv named students.csv the file contain following columns

- student_id, name, maths, science, english
- S001, Kamal, 78, 65, 80
- S002, Nimal, 45, 55, 60
- S003, Amal, 90, 88, 95

a. Write a python program using csv module to:
    1. Read all the student records from the csv file
    2. Calculate a total marks and the avg marks of each student
    3. Determine whether each student is passed or failed (Pass Marks for each subject is 50)
    4. Display students whose avg marks are above 75
    5. Save fail student details in to another csv file named -> failed_students.csv

```python
import csv

failed_std = []
avg_scoroing_std = []

with open('students.csv', 'r', newline='') as file:
    data = csv.DictReader(file)
    for line in data:
        std_name = line["name"]
        maths = int(line["maths"])
        science = int(line["science"])
        english = int(line["english"])
        
        total_marks = maths + science + english
        avg_marks = total_marks / 3
        
        if maths >= 50 and science >= 50 and english >= 50: 
            print(f"{std_name} has passed every Subject has been passed")
        else: 
            failed_std.append(line)
            print(f"This student: {std_name} has failed to pass subject(s)")
        
        if avg_marks > 75: 
            avg_scoroing_std.append(std_name)
        
with open('failed_students.csv', 'w', newline='') as file:
    writer = csv.DictWriter(file, fieldnames=['student_id','name', 'maths', 'science', 'english'])
    writer.writeheader()
    writer.writerows(failed_std)

print(f"Passed students: {avg_scoroing_std}")
print(failed_std)
```
