Q. Write a normalize_scores that accepts a scaling_factor parameter (default value = 1.0) and arbitary number of scores. 
   Test the function using different input values and explain the outputs produce for each function call.

```python
def normalize_function(*scores, scaling_factor=1.0):
    return [score*scaling_factor for score in scores]

print(normalize_function(2, 80, 90, 70, 5.0))
print(normalize_function(100, 85, scaling_factor=0.5))
print(normalize_function())
```

Q. Write a python function transform_sequence that accept
 - An arbitary number of integers
 - A keyword only parameter mode with default val "linear"
 - Rules:
    a. You're function should transform each number based on the selected mode
        i. Linear Mode: f(x) = 2 * x - 3
        ii. Quadratic Mode: f(x) = x ** 2 - x
        iii. Factorial_sum Mode: For each number x compute sum of factorial 1 to x (Assume x >= 1): f(x) = 1! + 2! + 3! + ... + x!
    b. Function output should return a dictionary as below
       {
        "transformed": [...], # List of transformed values
        "total": ..., # sum of transformed values
        "largest": ..., # largest of transformed value
        "max": ..., # max of transformed values
        "min": ..., # min of transformed values
       } 
    c. Test cases:
       transform_sequence(1, 2, 3, 4) 
       transform_sequence(1, 2, 3, mode="quadratic") 
       transform_sequence(3, 4, 5, mode="factorial_sum") 

```python
def transform_sequence(*values, mode="linear"):
    transformed = []
    fact = 1
    match mode:
        case "linear": 
            transformed = [2 * val - 3 for val in values]
        case "quadratic":
            transformed = [val ** 2 - val for val in values]
        case "factorial_sum":
            transformed = [[(total_sum := total_sum + (fact := fact * i)) for i in range(1, x + 1)][-1] for x in values if (total_sum := 0) == 0 and (fact := 1) == 1]
        case _:
            return "Define a proper mode"
            
    
    return {
        "transformed": transformed,
        "total": sum(transformed),
        "min": min(transformed),
        "max": max(transformed),
    } 
    
    
transform_sequence(1, 2, 3, 4) 
transform_sequence(1, 2, 3, mode="quadratic") 
transform_sequence(3, 4, 5, mode="factorial_sum")
# transform_sequence(3, 4, 5, mode="khjbh")
```

Q. Write a function that calculates bill that accepts:
 - A required keyword argument customer (string)
 - Arbitary keyword arguments (**items) representing purchased items and their prices. (e.g, burger=5. fries=2.5, cola=1.5)
 
 Task:
    i. Compute the sub total (SUM OF ALL ITEM PRICES)
    ii. Apply a discount rule system based on kwargs:
        - If subtotal >= 50 -> 10% discount
        - If subtotal >= 100 -> 20% discount
    iii. Add service tax of 5% after discount

    Return Format
    {
        "customer":
        "subtotal":
        "discount":
        "tax":
        "total":
    }

    calculate_bill(customer="Kamal", burger=20, pizza=30, cola=5)

```python
def calculate_bill(customer, **bill_details):
    sub_total = total = discount = tax = 0
    
    for key, value in bill_details.items():
        if value < 0: return "Enter a proper price amount!"
        else: sub_total += float(value) 
                
    discount = sub_total * 0.1 if sub_total >= 50 else (sub_total * 0.2 if sub_total >= 100 else 0.0)
    
    tax = discount * 0.05
    total = sub_total + discount + tax    
    return {
        "customer": customer,
        "subtotal": sub_total,
        "discount": discount,
        "tax": round(tax, 2), 
        "total": round(total, 2),
    }
            
print(calculate_bill(customer="Kamal", burger=20, pizza=30, cola=5))
```
